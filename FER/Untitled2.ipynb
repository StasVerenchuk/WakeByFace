```python
import os
import cv2
import numpy as np
from tensorflow.keras.models import load_model
```

```python
# Завантаження моделі
model = load_model("5classes_best_model.h5")

# Мітки класів (у тому ж порядку, в якому модель навчалась)
emotion_labels = ['Angry', 'Happy', 'Neutral', 'Sad', 'Surprise',]

# Шлях до папки з вашими зображеннями
custom_test_dir = "custom-test"  # онови шлях, якщо потрібно

# Прохід по кожному зображенню
for filename in os.listdir(custom_test_dir):
    if filename.lower().endswith(".jpg"):
        path = os.path.join(custom_test_dir, filename)

        # Завантаження та обробка зображення
        img = cv2.imread(path)
        gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
        resized = cv2.resize(gray, (48, 48))           # важливо: модель працює на 48x48
        normalized = resized.astype("float32") / 255.0
        input_img = np.expand_dims(normalized, axis=(0, -1))  # (1, 48, 48, 1)

        # Передбачення
        prediction = model.predict(input_img, verbose=0)
        label_index = np.argmax(prediction)
        label = emotion_labels[label_index]

        print(f"{filename}: {label}")
```
