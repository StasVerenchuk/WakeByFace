```python
from PIL import Image
```

```python
img = Image.open("custom-test/001_h.jpg")
img = img.resize((48, 48), Image.LANCZOS)
img.save("custom-test/output.jpg")
```

```python
img = Image.open("custom-test/output.jpg")
img = img.resize((224, 224), Image.LANCZOS)
img.save("custom-test/output_224.jpg")
```

```python
img = Image.open("custom-test/001_h.jpg").convert("RGB")

while img.size[0] // 2 >= 48 and img.size[1] // 2 >= 48:
    img = img.resize((img.size[0] // 2, img.size[1] // 2), Image.LANCZOS)

img = img.resize((48, 48), Image.LANCZOS)
img.save("custom-test/output_48.jpg")
```
